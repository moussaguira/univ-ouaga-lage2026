```python
import os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import geopandas as gpd
import flopy
```

```python
cwd = os.getcwd()

# Create a 100x100 array of random integers from 0 to 15
data = np.random.randint(0, 16, size=(100, 100))

# Plot the array
plt.imshow(data, cmap='viridis', interpolation='nearest')
plt.colorbar(label='Value')
plt.title('100x100 Random Integer Grid (0–15)')
plt.show()
figname = os.path.join(cwd, '../figures/testfig.pdf')
plt.savefig(figname)  
```
